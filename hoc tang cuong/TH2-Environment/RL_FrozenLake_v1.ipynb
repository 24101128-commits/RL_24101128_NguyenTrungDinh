```python
import random
import numpy as np
import gymnasium as gym
import matplotlib.pyplot as plt
from IPython import display
```

Part A : Environment Introduction


```python
env=gym.make("FrozenLake-v1",map_name="8x8",render_mode="rgb_array", is_slippery=False)
```

```python
env.reset()
img=env.render()
plt.imshow(img)
plt.show()
#Use size 8x8
```

```python
n_states=env.observation_space.n
n_actions=env.action_space.n
```

```python
print(f" num of states:{n_states}\n num of actions:{n_actions}")
```

```python
state,info=env.reset()
img=plt.imshow(env.render())
while True:
    action = env.action_space.sample()
    state, reward, terminated, truncated, info = env.step(action)
    img.set_data(env.render())
    display.display(plt.gcf())
    display.clear_output(wait=True)

    if terminated:
            break
```

```python
Q=np.zeros([n_states,n_actions])
Q.shape
Q
```

```python
episodes=1000
alpha=0.5
gamma=0.9
G=0
```

```python
for episode in range(1,episodes+1):
  state=env.reset()[0]
  done=False
  G=0
  while not done:

    if np.max(Q[state]) > 0:
        action = np.argmax(Q[state])

    else:
        action = env.action_space.sample()

    new_state,reward,done,info,a=env.step(action)
    Q[state,action]+=alpha*(reward+gamma*np.max(Q[new_state])-Q[state,action])
    G+=reward
    state=new_state
  if episode%50==0:
      print(f"episode {episode} sum of  reward :{G}")
```
